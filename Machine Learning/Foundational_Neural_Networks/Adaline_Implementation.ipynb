```python
from sklearn import datasets
import pandas as pd
import numpy as np
iris = datasets.load_iris()
f_names = iris.feature_names
df = pd.DataFrame(data = iris.data, columns = f_names)
df['target'] = iris.target
for i in range(len(df['target'])):
    if df.at[i, 'target'] == 2:
        df.at[i, 'target'] = 1 ##iris-virginica
    else:
        df.at[i, 'target'] = -1 ##not iris-virginica
df
```

```python
train_data = df[5:145] #large dataset to train the algorithm on
train_index = []
for x in range(140):
    train_index.append(x)
train_data.index = train_index
train_data
```

**--Q8(10 points)--**
"""Based on the discussion in class to implement your own non-OOP version Adaline algorithm, it can be a function accepts a dataframe, a learning rate, number of iterations and return the final SSE, learned weights and bias
You are allowed to implement your function based on author's OOP implementation. 
"""

```python
def Adaline(data, rate, epoch):
    rgen = np.random.RandomState(1)
    weights = rgen.normal(loc = 0.0, scale = 0.01, size = data.shape[1])
    
    for i in range(epoch):
        predicted_output = np.dot(data[['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)']], weights[:4]) + weights[4]
        error = []
        SSE = 0
        for j in range(len(predicted_output)):
            error.append(data.at[j, 'target'] - predicted_output[j])
            SSE += ((data.at[j, 'target'] - predicted_output[j]) ** 2) / 4
            if SSE == 0:
                break
        weights[:4] += rate * np.dot(data[['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)']].transpose(), error)
        weights[4] += rate * sum(error)
        print(SSE)
        if SSE == 0:
            break
    
    bias = weights[4]
        
    return SSE, weights[:4], bias
```

**--Q9--(3 points)**
implement a function named predict that accept a sample, learned weights and bias to predict the output. If the calculated weighted sum is greater than threshold, return output 1, otherwise, return output -1

```python
def Predict(data, weights, bias, threshold):
    results = []
    predict = np.dot(data[['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)']], weights) + bias
    for j in range(len(predict)):
        results.append(1 if predict[j] > threshold else -1)
    
    return results
```

```python
sse, l_weights, l_bias = Adaline(train_data, .0001, 100000)
print('Weights:', l_weights, '| Bias:', l_bias, '| Final SSE:', sse)
```

```python
outcome = Predict(test_data, l_weights, l_bias, 0)
error = 0
for x in range(len(outcome)):
    print('Prediction:', 'iris-virginica' if outcome[x] == 1 else 'not iris-virginica', ":", 'Target:',  'iris-virginica' if test_data.at[x, 'target'] == 1 else 'not iris-virginica')
    if outcome[x] != test_data.at[x, 'target']:
        error += 1
print('Errors:', error)
```
