```python
import numpy as np
import pandas as pd
mat = ([[1, 2.5, 3.5, 0], [1, 4.5, 2.5, 1], [1, 1.5, 2.0, 0], [1, 3.5, 4.0, 1]])
w = [0, 0, 0]
thresh = 0.4
learning_rate = 0.1
for j in range(10):
    error = 0
    predicted_outputs = []
    for sample in mat:
        weighted_sum = sample[1] * w[1] + sample[2] * w[2] + w[0]
        predicted_output = (1 if weighted_sum > thresh else 0)
        predicted_outputs.append(predicted_output)
        if predicted_output != sample[3]:
            error +=1
        for i in range(len(w)):
            w[i] = w[i] + learning_rate * (sample[3] - predicted_output) * sample[i]
    print('Run', j, ': Predictions:', predicted_outputs, '; Targets: [0, 1, 0, 1] ; Errors:', error)
    if error == 0:
        break
    print('Current Weights:')
    for y in range(len(w)):
        print('w[', y, ']', w[y])
print('Predicted outcomes line up with targets. Final weights:')
for x in range(len(w)):
    print('w[', x, ']', w[x])
```
